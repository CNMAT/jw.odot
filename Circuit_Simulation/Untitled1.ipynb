```python
import numpy as np
import matplotlib.pyplot as plt

SR = 48000
T = 1/SR
omega=np.linspace(0, np.pi,SR)
z = np.exp(complex(0,1)*omega)
warped_z=2/T * np.tan(z*T/2)
s = z / np.tan(z*T/2) * (z-1)/(z+1)
R1=10000
R2=100000
R3=220
R4=3300
C1=10e-6
C2=0.033e-6
C3=0.12e-6

N=s*C1 / (1+s*C1*R1)
D=s*C1/(1+s*C1*R1) + 1/R2 + s*C2/(1+s*C2*R2) + s*C3/(1+s*C3*R4)

H = N / D

plt.plot(omega*SR / (2*np.pi), np.abs(H))
plt.xscale('log')
plt.show()
plt.plot(omega, 180*np.arctan2(H.imag,H.real)/np.pi)
plt.xscale('log')
plt.show()
```
