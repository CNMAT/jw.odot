## Deriving Time Domain Representation of Filters by Tustin's Method ##


### Active inverting Low Pass Filter ###
![Fig.1: Active Inverting Low Pass Filter](![Active_Lowpass_Filter_RC.svg.png] "Optional title text")

Let us consider the transfer function of an active first-order inverting amplifier. This is made with an operational amplifier with a grounded positive input and negative feedback. The voltage in, $v_{in}$, arrives at the inverting input of the opamp through a resistor $R_1$. The inverting input also accepts feedback from the opamp output through resistor $R_2$ and capacitor $C$, arranged in parallel. Using Kirchoff's Current law, one can derive the voltages at each node in the network. Considering the node corresponding to the inverting input of the opamp, $v_{-}$, gives the following relation:

$$ \frac{v_{in} - v_-}{R_1} + \frac{v_{out} - v_-}{R_2} + \frac{v_{out} - v_-}{\frac{1}{sC}} = 0$$

Rearranging, we get 

$$ v_{out}\left(1+sCR_2\right) = v_-\left(\frac{R_2}{R_1} + 1 + sCR_2\right) - \frac{R_2}{R_1}v_{in}$$

Given that the non-inverting opamp input, $v_+$ is connected to ground ($0V$), and in this feedback arrangement, the operational amplifier will do all that it can to minimize the difference between its two inputs, we see that $v_- \rightarrow 0V$. Therefore, we have

$$ v_{out}\left(1+sCR_2\right) =  - \frac{R_2}{R_1}v_{in}$$

...which gives the transfer function

$$H(s) = \frac{Y(s)}{X(s)} \equiv \frac{v_{out}}{v_{in}} = \frac{\frac{-R_2} {R_1}}{(1+sCR_2)} $$

## Tustin's Method ##
The above equation exists in the continuous (Laplace) domain. To convert to the z-domain we simply replace $s$:

$$s = \frac{2}{T}\frac{1-z^{-1}}{1+z^{-1}}$$

where $T$ is the sampling period, defined as the inverse of the sampling frequency, $f_s$, by 

$$T = 1 / f_s$$.

For the above filter definition, this gives 

$$\frac{Y(s)}{X(s)} = \frac{\frac{-R_2}{R_1} } {1+\frac{2}{T}\frac{1-z^{-1}}{1+z^{-1}}R_2C}$$

Rearranging, we have

$$X(s)\frac{-R_2 T}{R_1}(1+z^{-1}) = Y(s)\left[ \left(T+2R_2C\right) + \left(T-2R_2C\right)z^{-1} \right]$$

We can now transform from the z-domain to the time-domain to find

$$\frac{-R_2 T}{R_1}\left(x[n] + x[n-1]\right) = \left(T+2R_2C\right) y[n] + \left(T-2R_2C\right)y[n-1]$$

Rearranging, we finally find:

$$y[n] = \frac{-R_2T}{R_1T+2R_1R_2C}\left[x[n] + x[n-1] \right] - \frac{T-2R_2C}{T+2R_2C}y[n-1]$$

More compactly, we introduce coefficients to simplify the picture

$$y[n] = b_0x[n] + b_1x[n-1] - a_0y[n-1]$$

where 

$$b_0 = b_1 = \frac{-R_2T}{R_1T+2R_1R_2C}$$

and

$$a_0 = \frac{T-2R_2C}{T+2R_2C}$$
