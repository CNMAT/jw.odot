## Deriving Time Domain Representation of Filters by Tustin's Method ##

### Passive Low Pass Filter ###
In the simplest case, a filter is simply a resistance and a reactive impedance (capacitance or inductance) in some arrangement. Let's begin with the passive Low Pass Filter, which is acheived with a single resistor and capacitor in a "L" configuration.

<img src="Passive_LPF.webp" style="background-color: #FFFFFF" alt="LowPass Filter" width="200"  />

In this arrangement we have the input $v_{in}$ passing through a resistor, $R$, connected to a capacitor bridging $v_{out}$ with ground. In this configuration, we can simply consider the voltage at the $v_{out}$ node. By Kirchoff's Current Law (KCL) we know that the sum of the currents into/out of a circuit node must be zero.

$$\sum_{n=0}^N I_n = 0$$

In this case, we need only consider the currents passing through the components $R$ & $C$ to characterize the circuit's behavior at $v_{out}$.  We calculate the sum of currents using Ohm's Law, $V=IR$, which gives $I = \frac{V}{R}$. K

Thus,

$$\frac{v_{in}-v_{out}}{R} + \frac{v_{ground} - v_{out}}{\frac{1}{sC}} = 0$$

Here, we use the fact that the impedance of a capacitor is defined as

$$Z_{capacitor} = \frac{1}{sC}$$

Simplifying Eq.1 with the assumption that $v_{ground}=0$, we find

$$\frac{v_{in}}{R}=\frac{v_{out}}{R} + sCv_{out}$$

$$v_{in} = v_{out}(1+sRC)$$

$$H(s) = \frac{v_{out}}{v_{in}} = \frac{1}{1+sRC}$$

Eq.2 is now in the form that we recognize as a transfer function $\left(H(s) = \frac{output}{input}\right)$.

In the s-domain (The Laplace Domain) the impedance is a function of the complex variable $s=\sigma + i\omega$. We wish to obtain a discrete time-domain representation of $H(s)$ that might be suitable for implementation in a digital system or software. For this, we use the Bilinear Transform (Tustin's Method), which maps the s-plane to the z-plane. (In terms of a geometric representation, the bilinear transform maps the points in positive real part of the s-plane onto the region outside the unit circle in the z-plane; the negative s-plane is mapped to points inside the unit circle). Tustin's Method achieves this transformation (to first approximation) by substitution of the variable, $s$.

$$s = \frac{1}{T} ln(z) $$

However, in this form the resulting expressions can be difficult to manipulate, so the common practice is to use a Taylor Expansion of $ln(z)$ in the following form.

$$s = \frac{2}{T}\left[\frac{z-1}{z+1} + \frac{1}{3}\left(\frac{z-1}{z+1}\right)^3 + \frac{1}{5}\left(\frac{z-1}{z+1}\right)^5 + \frac{1}{7}\left(\frac{z-1}{z+1}\right)^7 + \dots \right]$$

Where the sampling period is the inverse of the sampling rate $\left(T \equiv 1/f_s \right)$.

Approximating to first order, we have

$$s \approx \frac{2}{T} \frac{z-1}{z+1}$$

Or more conveniently

$$s \leftarrow \frac{2}{T} \frac{(1-z^{-1})}{(1+z^{-1})}$$

Performing this substitution in Eq.XXX, we now have

$$H(z) = \frac{1}{1+RC\frac{2}{T} \frac{(1-z^{-1})}{(1+z^{-1})}}$$

Simplifying, we have

$$H(z) = \frac{T(1+z^{-1})}{T(1+z^{-1})+2RC(1-z^{-1})}$$

This is the discrete time transfer function in the z-plane. We wish to represent this in terms of an input $x[n]$ and output $y[n]$, so we consider $H(z)$ as the ratio of the z-transforms of these functions $X(z)$ and $Y(z)$.

$$\frac{Y(z)}{X(z)}=\frac{T(1+z^{-1})}{T(1+z^{-1})+2RC(1-z^{-1})}$$

From which we can now perform some basic algebraic manipulation to arrive at 

$$(T+2RC) Y(z) + (T-2RC) Y(z) z^{-1} = T X(z) + TX(z) z^{-1}$$

At this point, it is convenient to take the inverse z-transform. For now, it is sufficient to know the following z-transforms

| Time Domain         | z-Transform         |
|---------------------|---------------------|
| $x[n]$              | $X(z)$              |
| $x[n-k]$            | $X(z)z^{-k}$        |
| $a_1x[n]+a_2x[n]$   | $a_1X(z)+a_2X(z)$   |
|---------------------|---------------------|

Using just these transforms we arrive at the following

$$y[n] = b_0  x[n] + b_1  x[n-1] - a_0  y[n-1]$$

where

$$b_0 = b_1 = \frac{T}{T+2RC}$$

$$a_0 = \frac{T-2RC}{T+2RC}$$





### Active inverting Low Pass Filter ###
<img src="Active_Lowpass_Filter_RC.png" style="background-color: #FFFFFF" alt="LowPass Filter" width="200"  />



Let us consider the transfer function of an active first-order inverting amplifier. This is made with an operational amplifier with a grounded positive input and negative feedback. The voltage in, $v_{in}$, arrives at the inverting input of the opamp through a resistor $R_1$. The inverting input also accepts feedback from the opamp output through resistor $R_2$ and capacitor $C$, arranged in parallel. Using Kirchoff's Current law, one can derive the voltages at each node in the network. Considering the node corresponding to the inverting input of the opamp, $v_{-}$, gives the following relation:

$$ \frac{v_{in} - v_-}{R_1} + \frac{v_{out} - v_-}{R_2} + \frac{v_{out} - v_-}{\frac{1}{sC}} = 0$$

Rearranging, we get 

$$ v_{out}\left(1+sCR_2\right) = v_-\left(\frac{R_2}{R_1} + 1 + sCR_2\right) - \frac{R_2}{R_1}v_{in}$$

Given that the non-inverting opamp input, $v_+$ is connected to ground ($0V$), and in this feedback arrangement, the operational amplifier will do all that it can to minimize the difference between its two inputs, we see that $v_- \rightarrow 0V$. Therefore, we have

$$ v_{out}\left(1+sCR_2\right) =  - \frac{R_2}{R_1}v_{in}$$

...which gives the transfer function

$$H(s) = \frac{Y(s)}{X(s)} \equiv \frac{v_{out}}{v_{in}} = \frac{\frac{-R_2} {R_1}}{(1+sCR_2)} $$

## Tustin's Method ##
The above equation exists in the continuous (Laplace) domain. To convert to the z-domain we simply replace $s$:

$$s = \frac{2}{T}\frac{1-z^{-1}}{1+z^{-1}}$$

where $T$ is the sampling period, defined as the inverse of the sampling frequency, $f_s$, by 

$$T = 1 / f_s$$.

For the above filter definition, this gives 

$$\frac{Y(s)}{X(s)} = \frac{\frac{-R_2}{R_1} } {1+\frac{2}{T}\frac{1-z^{-1}}{1+z^{-1}}R_2C}$$

Rearranging, we have

$$X(s)\frac{-R_2 T}{R_1}(1+z^{-1}) = Y(s)\left[ \left(T+2R_2C\right) + \left(T-2R_2C\right)z^{-1} \right]$$

We can now transform from the z-domain to the time-domain to find

$$\frac{-R_2 T}{R_1}\left(x[n] + x[n-1]\right) = \left(T+2R_2C\right) y[n] + \left(T-2R_2C\right)y[n-1]$$

Rearranging, we finally find:

$$y[n] = \frac{-R_2T}{R_1T+2R_1R_2C}\left[x[n] + x[n-1] \right] - \frac{T-2R_2C}{T+2R_2C}y[n-1]$$

More compactly, we introduce coefficients to simplify the picture

$$y[n] = b_0x[n] + b_1x[n-1] - a_0y[n-1]$$

where 

$$b_0 = b_1 = \frac{-R_2T}{R_1T+2R_1R_2C}$$

and

$$a_0 = \frac{T-2R_2C}{T+2R_2C}$$

### Active inverting High Pass Filter ###
<img src="Active_high-pass_filter_(I_order).svg.png" style="background-color: #FFFFFF" alt="High Pass Filter" width="200"  />

Here, similarly, we can use Kirchoff's Current law to obtain the currents flowing into the $v_-$ inverting input.

$$\frac{(v_{in}-v_-)}{\frac{1}{sC} + R_1} + \frac{v_{out}-v_-}{R_2} = 0$$

Once again, noting that the non-inverting input is grounded, the opamp will do everything it can to keep $v_- \rightarrow 0V$, so to first approximation those terms can be omitted. Rearranging, this yields


$$H(s) = \frac{v_{out}}{v_{in}} = \frac{-sR_2C}{1+sR_1C}$$

We can now obtain the z-transform through application of Tustin's method:

$$H(z) = \frac{Y(z)}{X(z)} = \frac{-2R_2C(1-z^{-1})}{T(1+z^{-1}) + 2R_1C(1-z^{-1})}$$

which leads to 

$$Y(z)\left[T(1+z^{-1}) + 2R_1C(1-z^{-1})\right] = X(z)\left[-2R_2C(1-z^{-1})\right]$$

Performing the inverse z-transform we get the discrete time-domain representation of

$$Y(z)  = \frac{-2R_2C}{(T+2R_1C)}\left[X(z) - X(z)z^{-1}\right] - \frac{(T-2R_1C)}{(T+2R_1C)}Y(z)z^{-1}$$

becomes

$$y[n] = \frac{-2R_2C}{(T+2R_1C)}\left(x[n] - x[n-1]\right)- \frac{(T-2R_1C)}{(T+2R_1C)}y[n-1]$$

or

$$y[n] = b_0x[n] + b_1x[n-1] - a_0y[n-1]$$

where

$$a_0 = \frac{(T-2R_1C)}{(T+2R_1C)}$$

and

$$b_0 = b_1 = \frac{2R_2C}{\left(T+2R_1C\right)}$$
