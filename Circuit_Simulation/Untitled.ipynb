```python
import numpy as np
from scipy import signal as sig
import matplotlib.pyplot as plt

N = 4096
f=10
d2=1010
x = np.linspace(0,N-1, N)
y1 = np.exp(-7*x/N)*np.sin(x*f*2*np.pi/N)
y2 = 0.5*np.exp(-7*np.maximum(x-d2,0)/N)*np.sin(np.maximum(x-d2,0)*f*2*np.pi/N)
# plt.plot(x,y1)
# plt.plot(x,y2)
y_sum = y1+y2
plt.plot(x,y_sum)
plt.show()
```

```python
h = sig.hilbert(y_sum)
plt.plot(x,np.abs(h))
plt.show()
```
